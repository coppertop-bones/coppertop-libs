```python
from coppertop.dm.utils.p9 import PP
from plotnine import ggplot, aes, geom_bar, geom_point, labs
```

```python
from plotnine.data import mpg
```

```python
(ggplot(mpg)
    + aes(x='class')
    + geom_bar(size=20)
 ) >> PP;
```

```python
(ggplot(mpg)
    + aes(x='displ', y='hwy', color='class')
    + geom_point()
    + labs(title='Engine Displacement vs. Highway Miles per Gallon', x='Engine Displacement, in Litres', y='Highway Miles per Gallon')
) >> PP;
```
